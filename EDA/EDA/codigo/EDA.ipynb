## CREAR VARIABLE OBJETIVO FINANCIERA (ROI)
- APARTIR DE LOS DATOS OBTENIDOS EN LAS TABLAS GAME PODEMOS DETERMINAR LAS VARIABLES OBJETIVO PARA DETERMINAR POSTERIORMENTE SI UN
- JUGADOR FUE RENTABLE O NO
- Objetivo

Crear una tabla que diga:

cuántos partidos jugó cada jugador

cuántos puntos produjo

cuántos años duró su carrera

y si fue rentable o no

```python
# Convertir directamente la fecha en game

game["game_date"] = pd.to_datetime(game["game_date"])

# Ver resultado

game.head()
```

```python
# vARIABLE 2. CREACIÓN DE UN ROI BASADO EN LA DURACIÓN DE LA CARRERA 

roi = players.groupby("player_id").agg(

    año_inicio=("from_year", "min"),

    año_fin=("to_year", "max"),

    experiencia=("season_exp_calculated", "max")

).reset_index()


# años carrera

roi["años_carrera"] = roi["año_fin"] - roi["año_inicio"]


# variable rentable

roi["rentable"] = (

    roi["experiencia"] >= 5

).astype(int)


roi.head()
```

```python
# Grafica rentable vs no rentable

plt.figure(figsize=(4,2))

sns.countplot(
    data=roi,
    x="rentable"
)

plt.title("Cantidad de jugadores rentables vs no rentables")

plt.xlabel("Rentable (0 = No, 1 = Sí)")

plt.ylabel("Cantidad de jugadores")

plt.show()
```

```python
#DISTRIBUCIÓN DE AÑOS DE CARRERA 

plt.figure(figsize=(5,2))

sns.histplot(
    roi["años_carrera"],
    bins=20
)

plt.title("Distribución de años de carrera")

plt.xlabel("Años de carrera")

plt.ylabel("Cantidad de jugadores")

plt.show()
```

```python
# EXPERIENCIA VS RENTABILIDAD 

promedio = roi.groupby("rentable")["experiencia"].mean()

plt.figure(figsize=(4,2))

promedio.plot(kind="bar")

plt.title("Promedio de experiencia según rentabilidad")

plt.xlabel("Rentable (0 = No, 1 = Sí)")

plt.ylabel("Años promedio de experiencia")

plt.show()
```

## METRICAS DE EFICIENCIA


```python
metricas = pd.DataFrame()

metricas["game_id"] = game["game_id"]

# Puntos totales : (Puntos): El total de puntos anotados.

metricas["PTS"] = game["pts_home"] + game["pts_away"]

# Tiros de Campo: Field Goals % (Porcentaje de tiros de campo): Relación entre tiros intentados y convertidos.

metricas["FG_pct"] = (

    (game["fgm_home"] + game["fgm_away"])

    /

    (game["fga_home"] + game["fga_away"])

)

# Tiros libres: Free Throws % (Porcentaje de tiros libres): Clave para medir la técnica y el rendimiento bajo presión.

metricas["FT_pct"] = (

    (game["ftm_home"] + game["ftm_away"])

    /

    (game["fta_home"] + game["fta_away"])

)

# 3P % : (Porcentaje de triples): Define si es un jugador perimetral o de pintura.

metricas["FG3_pct"] = (

    (game["fg3m_home"] + game["fg3m_away"])

    /

    (game["fg3a_home"] + game["fg3a_away"])

)

metricas.head()
```

```python
# crear una sola variable offensive score

metricas["efficiency_score"] = (

    metricas["FG_pct"]

    + metricas["FT_pct"]

    + metricas["FG3_pct"]

) / 3
```

```python
#RENDIMIENTO DE LA OFENSIVA TOTAL

metricas["volume_score"] = metricas["PTS"]

metricas["efficiency_score"] = (

    metricas["FG_pct"]

    + metricas["FT_pct"]

    + metricas["FG3_pct"]

) / 3


metricas["offensive_score"] = (

    metricas["volume_score"] * 0.7

    + metricas["efficiency_score"] * 100 * 0.3

)
```

```python
# ============================================
# CREAR EDAD DE ENTRADA A LA NBA
# ============================================

# convertir birthdate
players["birthdate"] = pd.to_datetime(players["birthdate"])


# extraer año nacimiento
players["birth_year"] = players["birthdate"].dt.year


# convertir season a numero
draft["season"] = pd.to_numeric(draft["season"], errors="coerce")


# unir draft con players
edad = draft.merge(

    players[["player_id", "birth_year"]],

    on="player_id",

    how="left"

)


# calcular edad entrada
edad["edad_entrada"] = edad["season"] - edad["birth_year"]

edad = edad.merge(

    roi[["player_id", "rentable"]],

    on="player_id",

    how="left"

)

edad = edad.merge(

    roi[["player_id", "rentable"]],

    on="player_id",

    how="left"

)

edad.head()
```

```python
# ======================================================
# CREAR TABLA EDAD + RENTABILIDAD
# ======================================================

# convertir birthdate
players["birthdate"] = pd.to_datetime(players["birthdate"], errors="coerce")

# extraer año nacimiento
players["birth_year"] = players["birthdate"].dt.year


# usar season como draft_year
draft["draft_year"] = pd.to_numeric(draft["season"], errors="coerce")


# crear tabla edad
edad = draft.merge(

    players[["player_id", "birth_year"]],

    on="player_id",

    how="left"

)


# calcular edad entrada NBA
edad["edad_entrada"] = edad["draft_year"] - edad["birth_year"]


# unir RENTABILIDAD (tabla roi que ya creaste antes)
edad = edad.merge(

    roi[["player_id", "rentable"]],

    on="player_id",

    how="left"

)

# ver resultado
edad.head()
```

```python
# ============================================
# LIMPIAR NULOS
# ============================================

edad_clean = edad.dropna(subset=["edad_entrada", "rentable"])


# ============================================
# CREAR RANGOS DE EDAD
# ============================================

edad_clean["rango_edad"] = pd.cut(

    edad_clean["edad_entrada"],

    bins=[0, 19, 21, 23, 40],

    labels=[

        "≤ 19",

        "20-21",

        "22-23",

        "≥ 24"

    ]

)


# ============================================
# CALCULAR PROBABILIDAD
# ============================================

prob_edad = edad_clean.groupby("rango_edad")["rentable"].mean().reset_index()


# ============================================
# GRAFICA SIMPLE
# ============================================

plt.figure(figsize=(8,5))

plt.bar(

    prob_edad["rango_edad"],

    prob_edad["rentable"]

)

plt.title("Probabilidad de ser rentable según edad de entrada")

plt.xlabel("Edad de entrada")

plt.ylabel("Probabilidad")

plt.show()
```
